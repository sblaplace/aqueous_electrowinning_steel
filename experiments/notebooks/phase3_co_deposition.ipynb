# Phase III — Co-deposition (Fe-Ni + C) + pulse coupling

```python
from models.co_deposition import build_phase3_model
import matplotlib.pyplot as plt
import numpy as np

model = build_phase3_model(mechanism_fe_ni="hydroxide_suppression")
print(model.run_at_current(100))
print("Pulsed:", model.run_at_current_pulsed(50,100,0.5,"pe")["alloy_kinetics"])

sweep = model.run_sweep(np.linspace(10,200,20))
plt.plot(sweep["j_mA_cm2"], sweep["fe_wt_percent"], label="Fe wt%")
plt.plot(sweep["j_mA_cm2"], sweep["ni_wt_percent"], label="Ni wt%")
plt.legend(); plt.xlabel("j mA/cm2"); plt.ylabel("wt%"); plt.grid(alpha=0.25); plt.show()
```
